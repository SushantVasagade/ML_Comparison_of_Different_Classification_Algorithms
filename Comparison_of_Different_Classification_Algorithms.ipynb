# Theory

## 1. Classification

Classification is a **Supervised Machine Learning technique** in which a model learns from labelled training data and predicts the class or category of new observations.

For example, in the Iris dataset, the model predicts whether a flower belongs to:

- Setosa
- Versicolor
- Virginica

General representation:

$$
X \rightarrow Y
$$

where:

- $X$ = Input features
- $Y$ = Class label

## 2. Logistic Regression

Logistic Regression is a supervised classification algorithm used to predict the probability of a class.

For binary classification, the sigmoid function is:

$$
P(Y=1|X) = \frac{1}{1+e^{-z}}
$$

where:

$$
z = \beta_0 + \beta_1X_1 + \beta_2X_2 + \cdots + \beta_nX_n
$$

For multiclass classification, Logistic Regression can use strategies such as **One-vs-Rest** or **multinomial classification**.

## 3. Decision Tree

A Decision Tree is a tree-based classification algorithm that makes decisions using a sequence of conditions.

The structure consists of:

- Root Node
- Decision Nodes
- Branches
- Leaf Nodes

A common measure for selecting a split is Gini impurity:

$$
Gini = 1 - \sum_{i=1}^{C} p_i^2
$$

where $p_i$ is the probability of class $i$.

## 4. Random Forest

Random Forest is an ensemble learning algorithm that combines multiple Decision Trees.

Each tree produces a prediction and the final classification is generally determined by majority voting.

Advantages:

- Reduces overfitting compared with a single tree.
- Works well with many types of data.
- Provides robust predictions.

## 5. Support Vector Machine (SVM)

Support Vector Machine is a supervised learning algorithm that finds a suitable decision boundary between different classes.

The important points closest to the decision boundary are called **Support Vectors**.

For a linear SVM, the decision boundary can be represented as:

$$
w^TX+b=0
$$

SVM can also handle nonlinear classification using kernels such as:

- Linear Kernel
- Polynomial Kernel
- RBF Kernel

## 6. K-Nearest Neighbors (KNN)

KNN is a distance-based classification algorithm.

It classifies a new data point based on the classes of its nearest neighbours.

The commonly used Euclidean distance is:

$$
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

The value of $K$ represents the number of neighbours considered for prediction.

## 8. Confusion Matrix

A confusion matrix is used to evaluate the performance of a classification model.

For a multiclass problem, it shows the number of:

- Correct predictions
- Incorrect predictions
- Actual classes
- Predicted classes

For each class, we can calculate:

- True Positive (TP)
- True Negative (TN)
- False Positive (FP)
- False Negative (FN)

## 9. Classification Performance Metrics

### Accuracy

Accuracy represents the proportion of correctly classified observations.

$$
Accuracy = \frac{TP+TN}{TP+TN+FP+FN}
$$

### Precision

Precision represents how many predicted positive observations are actually positive.

$$
Precision = \frac{TP}{TP+FP}
$$

### Recall

Recall represents how many actual positive observations are correctly identified.

$$
Recall = \frac{TP}{TP+FN}
$$

For the multiclass Iris dataset, **macro averaging** is used so that each class contributes equally to the final Precision and Recall values.

## STEP 1 — Import Required Libraries

In this step, the required Python libraries and Scikit-learn modules are imported.

### Libraries used

**NumPy**
- Used for numerical operations.

**Pandas**
- Used for creating and manipulating the dataset in DataFrame format.

**Matplotlib**
- Used for creating graphs and visualizations.

### Scikit-learn modules used

- `load_iris()` loads the Iris dataset.
- `train_test_split()` divides data into training and testing sets.
- `StandardScaler()` standardizes numerical features.
- Classification classes create the required Machine Learning models.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report
)
```

## STEP 2 — Load Iris Dataset

The Iris dataset is loaded using the `load_iris()` function from Scikit-learn.

The dataset contains measurements of Iris flowers belonging to three different species.

The `load_iris()` function returns a dataset object containing:

- Feature data
- Target values
- Feature names
- Target class names
- Dataset description

The loaded dataset is stored in the variable `iris` for further processing.

```python
iris = load_iris()

print("Iris dataset loaded successfully.")
```

## STEP 3 — Create a Pandas DataFrame

The numerical feature data from the Iris dataset is converted into a Pandas DataFrame.

The four original feature names are assigned as column names:

- Sepal length
- Sepal width

## STEP 13 — Separate Features and Target

Machine Learning classification requires separating the input features from the output target.

The four Iris measurements are used as input features:

- Sepal length
- Sepal width
- Petal length
- Petal width

The numerical `target` column is used as the output variable.

Therefore:

$$
X = Features
$$

and

$$
y = Target
$$

The feature matrix has 150 observations and 4 input features.

```python
X = data[iris.feature_names]

y = data["target"]

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)
```

## STEP 14 — Inspect Feature Data

The feature matrix `X` contains only the four input measurements.

The target and species columns are not included because they represent the output rather than input features.

Displaying the first five rows helps verify that the correct features have been selected.

```python
print("Input features:")
display(X.head())
```

## STEP 15 — Split Dataset into Training and Testing Sets

The dataset is divided into two parts:

### Training Dataset

The training data is used by the Machine Learning algorithms to learn patterns from the observations.

### Testing Dataset

The testing data is kept separate and used to evaluate how well the trained models perform on unseen observations.

In this experiment:

$$
Training\ Data = 80\%
$$

$$
Testing\ Data = 20\%
$$

`random_state=42` makes the split reproducible.

`stratify=y` maintains approximately the same class distribution in both training and testing datasets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

## STEP 16 — Create StandardScaler

Some Machine Learning algorithms are sensitive to the scale of input features.

In this experiment, feature standardization is required for:

- Logistic Regression
- Support Vector Machine
- K-Nearest Neighbors

`StandardScaler` transforms each feature using:

$$
z=\frac{x-\mu}{\sigma}
$$

where:

- $x$ = original feature value
- $\mu$ = feature mean
- $\sigma$ = feature standard deviation
- $z$ = standardized value

A scaler object is created before applying the transformation.

```python
scaler = StandardScaler()

print("StandardScaler object created successfully.")
```

## STEP 17 — Standardize Training and Testing Data

The scaler is fitted using only the training data.

This is important because the test data must remain unseen during the training process.

The following operations are performed:

### Training Data

`fit_transform()` calculates the required scaling parameters from the training data and transforms it.

### Testing Data

`transform()` uses the parameters learned from the training data to transform the test data.

This prevents information leakage from the testing dataset.

```python
X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training data standardized successfully.")
print("Testing data standardized successfully.")
```

## STEP 18 — Display Standardized Features

The standardized training data is converted back into a Pandas DataFrame for easier interpretation.

The original feature names are assigned to the standardized columns.

After standardization, the feature values are on a comparable scale.

This scaled data will be used by:

- Logistic Regression
- SVM
- KNN

```python
scaled_training_data = pd.DataFrame(
    X_train_scaled,
    columns=iris.feature_names
)

print("First five standardized training observations:")
display(scaled_training_data.head())
```

## STEP 19 — Create Logistic Regression Model

A Logistic Regression classifier is created using Scikit-learn.

The model is suitable for classification and can handle the three Iris classes.

The `max_iter` parameter specifies the maximum number of iterations allowed during model optimization.

A value of `500` is used to provide sufficient iterations for model convergence.

The model will later be trained using the standardized training features.

```python
logistic_regression_model = LogisticRegression(
    max_iter=500
)

print("Logistic Regression model created successfully.")
```

## STEP 20 — Create Decision Tree Model

A Decision Tree classifier is created.

The model classifies observations by learning decision rules from the training data.

The `random_state=42` parameter ensures reproducible results.

The `max_depth=5` parameter limits the maximum depth of the tree.

Limiting tree depth can help control excessive model complexity.

Decision Tree does not require feature standardization.

```python
decision_tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

print("Decision Tree model created successfully.")
```

## STEP 21 — Create Random Forest Model

A Random Forest classifier is created.

Random Forest combines multiple Decision Trees to produce a more robust prediction.

The `n_estimators=100` parameter specifies that 100 Decision Trees will be used.

The `random_state=42` parameter makes the experiment reproducible.

`n_jobs=1` is used so that the computation runs with a single worker, which is suitable for a basic laboratory setup and avoids unnecessary CPU usage.

Random Forest does not require feature standardization.

```python
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=1
)

print("Random Forest model created successfully.")
```

## STEP 22 — Create Support Vector Machine Model

A Support Vector Machine classifier is created using the `SVC` class.

The **RBF kernel** is selected because it can model nonlinear relationships between observations.

The `C` parameter controls the trade-off between classification error and model complexity.

In this experiment:

$$
C=1.0
$$

Feature scaling is used before training the SVM because SVM is sensitive to feature magnitude.

```python
support_vector_machine_model = SVC(
    kernel="rbf",
    C=1.0,
    random_state=42
)

print("Support Vector Machine model created successfully.")
```

## STEP 23 — Create K-Nearest Neighbors Model

A K-Nearest Neighbors classifier is created.

KNN classifies a new observation based on the classes of its nearest training observations.

The `n_neighbors` parameter specifies the number of neighbours considered.

In this experiment:

$$
K=5
$$

Because KNN uses distances between observations, standardized features are used for training and prediction.

```python
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

print("K-Nearest Neighbors model created successfully.")
```

## STEP 24 — Create Gaussian Naive Bayes Model

A Gaussian Naive Bayes classifier is created.

Gaussian Naive Bayes is suitable for continuous numerical features.

It assumes that the features approximately follow a Gaussian distribution within each class.

The model uses probability-based classification and does not require feature standardization for this experiment.

```python
gaussian_naive_bayes_model = GaussianNB()

print("Gaussian Naive Bayes model created successfully.")
```

## STEP 25 — Store All Classification Models

All six classification models are stored in a Python dictionary.

The dictionary provides a clear mapping between the model name and its corresponding model object.

The six models are:

1. Logistic Regression
2. Decision Tree
3. Random Forest
4. SVM
5. KNN
6. Gaussian Naive Bayes

This allows the models to be processed systematically during training and prediction.

```python
models = {
    "Logistic Regression": logistic_regression_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model,
    "SVM": support_vector_machine_model,
    "KNN": knn_model,
    "Naive Bayes": gaussian_naive_bayes_model
}

print("Classification models:")
for model_name in models.keys():
    print("-", model_name)
```

## STEP 26 — Train All Classification Models

Each classification model is trained using the training dataset.

Different models require different preprocessing:

### Standardized Features

The following algorithms use standardized training data:

- Logistic Regression
- SVM
- KNN

### Original Features

The following algorithms are trained using the original feature values:

- Decision Tree
- Random Forest
- Gaussian Naive Bayes

The `fit()` method is used to train each model using the corresponding training features and training target values.

```python
scaled_model_names = {
    "Logistic Regression",
    "SVM",
    "KNN"
}

for model_name, model in models.items():

    if model_name in scaled_model_names:
        model.fit(
            X_train_scaled,
            y_train
        )
    else:
        model.fit(
            X_train,
            y_train
        )

    print(
        model_name,
        "trained successfully."
    )
```

## STEP 27 — Generate Predictions

After training, each model is used to predict the target classes of the unseen test observations.

The `predict()` method generates the predicted class for each test observation.

For models trained using standardized data, `X_test_scaled` is used.

For models trained using original data, `X_test` is used.

The predictions of all six algorithms are stored in a dictionary for later evaluation.

```python
predictions = {}

for model_name, model in models.items():

    if model_name in scaled_model_names:
        predicted_classes = model.predict(
            X_test_scaled
        )
    else:
        predicted_classes = model.predict(
            X_test
        )

    predictions[model_name] = predicted_classes

    print(
        model_name,
        "predictions generated successfully."
    )
```

## STEP 28 — Calculate Accuracy, Precision and Recall

The predictions generated by each classification model are compared with the actual test labels.

Three important evaluation metrics are calculated.

### Accuracy

Measures the overall proportion of correctly classified observations.

$$
Accuracy=\frac{Correct\ Predictions}{Total\ Predictions}
$$

### Precision

Measures how many observations predicted as a particular class actually belong to that class.

$$
Precision=\frac{TP}{TP+FP}
$$

### Recall

Measures how many actual observations belonging to a class were correctly identified.

$$
Recall=\frac{TP}{TP+FN}
$$

Because the Iris dataset contains three classes, `average="macro"` is used for Precision and Recall.

Macro averaging gives equal importance to each of the three classes.

`zero_division=0` prevents an error if a class has no predicted positive observations.

The results are stored in a Pandas DataFrame and sorted by Accuracy.

```python
results = []

for model_name, predicted_classes in predictions.items():

    model_accuracy = accuracy_score(
        y_test,
        predicted_classes
    )

    model_precision = precision_score(
        y_test,
        predicted_classes,
        average="macro",
        zero_division=0
    )

    model_recall = recall_score(
        y_test,
        predicted_classes,
        average="macro",
        zero_division=0
    )

    results.append({
        "Algorithm": model_name,
        "Accuracy": model_accuracy,
        "Precision": model_precision,
        "Recall": model_recall
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("Classification performance comparison:")
display(results_df)
```

## STEP 29 — Generate Confusion Matrices

A confusion matrix is generated separately for every classification algorithm.

The rows represent the actual Iris species and the columns represent the predicted Iris species.

The three classes are:

- Setosa
- Versicolor
- Virginica

Correct predictions appear along the diagonal of the matrix.

Incorrect predictions appear outside the diagonal.

The confusion matrix helps identify which classes are being confused by a particular classification algorithm.

```python
for model_name, predicted_classes in predictions.items():

    confusion_matrix_values = confusion_matrix(
        y_test,
        predicted_classes
    )

    confusion_matrix_dataframe = pd.DataFrame(
        confusion_matrix_values,
        index=iris.target_names,
        columns=iris.target_names
    )

    print(
        f"\nConfusion Matrix — {model_name}"
    )

    display(confusion_matrix_dataframe)
```

## STEP 30 — Final Model Comparison and Classification Reports

The final step provides a complete comparison of all six classification algorithms.

The following operations are performed:

### 1. Performance Comparison

Accuracy, Precision, and Recall of all algorithms are displayed together.

### 2. Visualization

A bar chart is created to visually compare the three evaluation metrics.

### 3. Classification Reports

A detailed classification report is generated for every algorithm.

The classification report contains:

- Precision
- Recall
- F1-score
- Support

for each Iris species.

### 4. Identify Highest Test Accuracy

The algorithm having the highest **test Accuracy** is identified directly from the experimental results.

The result is not assumed beforehand; it is obtained from the actual predictions generated by the models.

```python
comparison_dataframe = results_df.set_index(
    "Algorithm"
)

plt.figure(figsize=(12, 6))

comparison_dataframe.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title(
    "Comparison of Classification Algorithms"
)

plt.xlabel(
    "Classification Algorithm"
)

plt.ylabel(
    "Performance Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=45
)

plt.legend(
    title="Evaluation Metrics"
)

plt.tight_layout()
plt.show()


for model_name, predicted_classes in predictions.items():

    print("=" * 70)
    print(
        f"Classification Report — {model_name}"
    )
    print("=" * 70)

    report = classification_report(
        y_test,
        predicted_classes,
        target_names=iris.target_names,
        zero_division=0
    )

    print(report)


highest_accuracy_index = results_df["Accuracy"].idxmax()

best_model_name = results_df.loc[
    highest_accuracy_index,
    "Algorithm"
]

highest_test_accuracy = results_df.loc[
    highest_accuracy_index,
    "Accuracy"
]

print("=" * 70)
print("FINAL CLASSIFICATION COMPARISON")
print("=" * 70)

display(results_df)

print(
    f"Algorithm with highest test accuracy: "
    f"{best_model_name}"
)

print(
    f"Highest test accuracy: "
    f"{highest_test_accuracy:.4f}"
)
```
